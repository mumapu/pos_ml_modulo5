# Etapa 2: Preparação da Base

Nesta etapa, vamos preparar as *features* (variáveis explicativas) para os modelos de Machine Learning. Isso envolve transformar variáveis categóricas (como estado civil, profissão) em valores numéricos (One-Hot Encoding) e separar os dados em treino e teste.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

import warnings
warnings.filterwarnings('ignore')
```

```python
# Carregando a base limpa da Etapa 1
df = pd.read_csv('../data/bank_cleaned.csv')
display(df.head())
```

## 1. Separação entre Features (X) e Target (y)

```python
X = df.drop(columns=['target'])
y = df['target']
```

## 2. Tratamento de Variáveis Categóricas (One-Hot Encoding)
Modelos de Machine Learning requerem entradas numéricas. Usaremos o `pd.get_dummies` para transformar as colunas de texto em colunas binárias (0 e 1).

```python
X_encoded = pd.get_dummies(X, drop_first=True)
print(f"Número de features após o encoding: {X_encoded.shape[1]}")
display(X_encoded.head())
```

## 3. Divisão em Treino e Teste
Para avaliar corretamente nossos baselines e modelos adaptativos, vamos separar a base para que a avaliação não seja enviesada.

```python
X_train, X_test, y_train, y_test = train_test_split(X_encoded, y, test_size=0.2, random_state=42, stratify=y)

print(f"Tamanho do Treino: {X_train.shape}")
print(f"Tamanho do Teste: {X_test.shape}")
```

## 4. Salvando os Dados Processados
Estes dados já estão prontos para a Etapa 3.

```python
X_train.to_csv('../data/X_train.csv', index=False)
X_test.to_csv('../data/X_test.csv', index=False)
y_train.to_csv('../data/y_train.csv', index=False)
y_test.to_csv('../data/y_test.csv', index=False)
print("Bases de treino e teste salvas com sucesso!")
```
